```python
import itertools
```

```python
templateFile = './slhaFiles/HeHe_500_100_SLHA.dat'
```

```python
with open(templateFile,'r') as f:
    template = f.readlines()
```

```python
mX = [400.,450.,500.,550.,600.,650.,700.,750.,800.,850.,900.,950.]
mY = [1.,100.,200.,300.]
lines_X = ['5000011 5.000000e+02 # mhe']
lines_Y = ['5000022 1.000000e+02 # ms0']
for x,y in itertools.product(mX,mY):
    newFile = templateFile.replace('500',str(int(x))).replace('100',str(int(y)))
    # newFile = newFile.replace('slhaFiles','new_slhaFiles')
    if newFile == templateFile:
        continue
    with open(newFile,'w') as f:
        for line in template:
            if line.strip() in lines_X:
                new_line = line.replace(' 5.000000e+02 #', f' {x:.6e} #')
            elif line.strip() in lines_Y:
                new_line = line.replace(' 1.000000e+02 #', f' {y:.6e} #')
            else:
                new_line = line
            f.write(new_line)
```
