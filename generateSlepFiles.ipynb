```python
import itertools
```

```python
templateFile = './slhaFiles/SlepSlep_500_1_SLHA.dat'
```

```python
with open(templateFile,'r') as f:
    template = f.readlines()
```

```python
mX = [750.,800.]
mY = [1.,10.,50.,100.,150.,200.,250.,300.,350.,400.,450.,500.]
lines_X = ['1000011 5.000000e+02 #', '2000011 5.000000e+02 #', '1000013 5.000000e+02 #', '2000013 5.000000e+02 #', '1000015 5.000000e+02 #', '2000015 5.000000e+02 #']
lines_Y = ['1000022 1.000000e+00 #']
for x,y in itertools.product(mX,mY):
    newFile = templateFile.replace('500',str(int(x))).replace('1',str(int(y)))
    newFile = newFile.replace('slhaFiles','new_slhaFiles')
    with open(newFile,'w') as f:
        for line in template:
            if line.strip() in lines_X:
                new_line = line.replace(' 5.000000e+02 #', f' {x:.6e} #')
            elif line.strip() in lines_Y:
                new_line = line.replace(' 1.000000e+00 #', f' {y:.6e} #')
            else:
                new_line = line
            f.write(new_line)
```
